```python
import wfdb, pandas as pd, io, urllib.request

DB = "auricle/1.0.0"
BASE = "https://physionet.org/files/auricle/1.0.0/"

# All recordings in the dataset
records = wfdb.get_record_list(DB)
print(len(records), "records")
print(records[:10], "...", records[-5:])

# Demographics and activity labels (small CSV files)
def read_csv(name):
    with urllib.request.urlopen(BASE + name) as r:
        return pd.read_csv(io.BytesIO(r.read()))

demo = read_csv("demographics.csv")
labels = read_csv("activity_labels.csv")
print("\nDEMOGRAPHICS:", demo.shape); print(demo.head(8).to_string())
print("\nCOLUMNS:", list(demo.columns))
print("\nACTIVITY LABELS:", labels.shape); print(labels.head(8).to_string())
```
